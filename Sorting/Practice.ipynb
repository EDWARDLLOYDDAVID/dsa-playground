```python
X = [5,3,8,5,6,7,2]

n = len(X)
print(X)

for i in range(n-1):
    for j in range(n-i-1):
        if X[j] > X[j+1]:
            X[j], X[j+1] = X[j+1], X[j]

print(X)
```

# Selection Sort

```python
mylist = [5,6,8,2,1]

n = len(mylist)
print(n,n-1)
for i in range(n-1):
  min_index = i
  for j in range(i+1, n):
     if mylist[j] < mylist[min_index]:
       min_index = j
  min_value = mylist.pop(min_index)
  mylist.insert(i, min_value)

print(mylist)
```
